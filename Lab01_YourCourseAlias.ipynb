## Step 1 | Analyze DNA | Minutes 16–25
### A. Run a baseline
This synthetic sequence is written 5′ to 3′ using the four standard DNA letters. We will count its length and the percentage of letters that are G or C. We are **not** identifying a real gene or predicting its function.

`dna` is a **variable**: a name attached to a value. The quoted letters are a **string**: text. `len(dna)` counts characters, and `.count("G")` counts uppercase G letters. A line beginning with `#` is a comment for a human reader.

Run the baseline cell. Leave it unchanged afterward.

````python
# Synthetic teaching sequence; use uppercase letters and no spaces.
dna = "ATGAAACCGGCTTTATAA"
length = len(dna)
gc_count = dna.count("G") + dna.count("C")
gc_percent = 100 * gc_count / length

print("Length (bases):", length)
print("G + C (bases):", gc_count)
print("GC (%):", round(gc_percent, 1))
````

## Step 2 | Simulate protein accumulation | Minutes 25–33
### A. Understand the model
Our invented cell starts with **0 arbitrary units of protein**. In each **one-minute step**, it loses **20% of the amount present at the beginning of the step**, and then receives **10 new units**.

The **model** is that rule and its assumptions. The **simulation** is the computer repeatedly applying the rule to calculate amounts over time. The printed values are calculated predictions, **not measurements from real cells**.

The amount is a continuous, relative quantity, not a count of individual molecules. Each step always represents one minute. The loss fraction is fixed at 0.20 per step; it is not a measured rate constant.

**Hand check:** After the first minute, the amount should be 10. After the next minute, 8 of those units remain and 10 new units are added.

The indented lines under `for` repeat. `range(1, 7)` means minutes 1 through 6; the endpoint 7 is excluded. Leave the indentation unchanged. `round` changes only what is displayed, not the unrounded running amount.

````python
# Each loop is one minute. Amounts are arbitrary protein units.
production = 10       # Units added in each one-minute step.
loss_fraction = 0.20  # Fraction of the existing pool lost each step.
protein = 0.0        # Starting amount; reset every time this cell runs.

print("Minute | Protein (arbitrary units)")
print(0, "|", protein)
for minute in range(1, 7):
    protein = protein * (1 - loss_fraction) + production
    print(minute, "|", round(protein, 2))
````

````python
# Each loop is one minute. Amounts are arbitrary protein units.
production = 20       # Units added in each one-minute step.
loss_fraction = 0.20  # Fraction of the existing pool lost each step.
protein = 0.0        # Starting amount; reset every time this cell runs.

print("Minute | Protein (arbitrary units)")
print(0, "|", protein)
for minute in range(1, 7):
    protein = protein * (1 - loss_fraction) + production
    print(minute, "|", round(protein, 2))
````
